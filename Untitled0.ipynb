```python
# ==========================================
# 1. INSTALASI LIBRARY TERBARU
# ==========================================
!pip install -q langchain langchain-groq langchain-community langchain-huggingface pypdf faiss-cpu

import os
from langchain_community.document_loaders import PyPDFDirectoryLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser

# ==========================================
# 2. KONFIGURASI API KEY GROQ
# ==========================================
os.environ["GROQ_API_KEY"] = "Masukkan KEY Anda"

# ==========================================
# 3. LOAD JURNAL PDF DARI FOLDER COLAB
# ==========================================
print("🔄 Membaca file PDF dari folder 'jurnal_finance' di Colab...")
loader = PyPDFDirectoryLoader("jurnal_finance")
docs = loader.load()
print(f"✅ Berhasil membaca {len(docs)} halaman dokumen.")

# ==========================================
# 4. TEXT SPLITTING (Chunking)
# ==========================================
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
chunks = text_splitter.split_documents(docs)
print(f"✂️ Dokumen dipecah menjadi {len(chunks)} chunks.")

# ==========================================
# 5. EMBEDDING & VECTOR STORE
# ==========================================
print("🧠 Membuat Vector Embeddings dengan HuggingFace...")
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
vector_store = FAISS.from_documents(chunks, embeddings)
retriever = vector_store.as_retriever(search_kwargs={"k": 5})
print("✅ Vector Store FAISS berhasil dibuat.")

# ==========================================
# 6. SETUP LLM GROQ & SYSTEM PROMPT (MODEL TERVERIFIKASI AKTIF)
# ==========================================
# Menggunakan openai/gpt-oss-20b yang terverifikasi aktif di infrastruktur Groq
llm = ChatGroq(model="openai/gpt-oss-20b", temperature=0.2)

template = """Anda adalah seorang profesor AI tingkat S2 yang ahli dalam mengevaluasi literatur akademik.
Gunakan dokumen konteks berikut untuk menjawab permintaan pengguna dengan akurat dan profesional.

Konteks Jurnal:
{context}

Pertanyaan:
{input}

Jawaban Akademis (Bahasa Indonesia):"""

prompt = ChatPromptTemplate.from_template(template)

# Fungsi pembantu untuk menggabungkan konten dokumen yang ditemukan oleh retriever
def format_docs(docs):
    return "\n\n".join([d.page_content for d in docs])

# ==========================================
# 7. MODEL LCEL PIPELINE
# ==========================================
rag_chain = (
    {"context": retriever | format_docs, "input": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)

query = """
Berdasarkan jurnal-jurnal keuangan dan UI/UX yang ada di dalam konteks dokumen, buatkan tabel "Literature GAP Matrix" dalam format Markdown.
Tabel harus memiliki kolom:
1. Peneliti & Tahun
2. Metode AI/LLM yang Digunakan
3. Domain / Fokus Evaluasi Finansial
4. Temuan Utama / Kelebihan
5. Keterbatasan / GAP (Celah Penelitian) yang Belum Diselesaikan

Di akhir tabel, berikan analisis mendalam (1-2 paragraf) bagaimana ide proyek saya yang berjudul "GenAI-Driven FinTech UX Assistant: Otomatisasi Evaluasi & Rekomendasi Antarmuka Mobile Banking menggunakan LLM Reasoning" hadir untuk mengisi GAP (celah penelitian) yang ditinggalkan oleh peneliti-peneliti terdahulu tersebut. Jawab dalam Bahasa Indonesia yang formal, berbobot, dan akademis standar S2.
"""

print("\n🚀 Menjalankan RAG Pipeline LCEL untuk menghasilkan GAP Matrix...")
response_text = rag_chain.invoke(query)

# ==========================================
# 8. OUTPUT HASIL & MENYIMPAN FILE
# ==========================================
print("\n=== HASIL LITERATURE GAP MATRIX ===\n")
print(response_text)

with open("Hasil_GAP_Matrix.md", "w", encoding="utf-8") as f:
    f.write(response_text)
print("\n💾 Hasil telah disimpan ke file 'Hasil_GAP_Matrix.md' di panel kiri Colab.")
```
